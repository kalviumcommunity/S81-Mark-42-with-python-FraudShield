# Running, Restarting, and Interrupting Jupyter Kernels

This notebook demonstrates kernel management: understanding kernel state, running cells in order, restarting, and interrupting execution.

## What is a Jupyter Kernel?

A **kernel** is the Python interpreter that runs your code. It:
- Executes cells when you run them
- **Remembers variables** from previously executed cells
- **Maintains state** until it is restarted
- Can be interrupted if execution is taking too long

Understanding kernel state is crucial for debugging and reproducibility.

```python
# Code Cell 1: Initialize Variables
# These variables will be remembered by the kernel

count = 0
data = []

print(f"✓ Kernel initialized")
print(f"  count = {count}")
print(f"  data = {data}")
```

```python
# Code Cell 2: Modify Variables
# The kernel REMEMBERS the variables from Cell 1
# If you hadn't run Cell 1, this would fail!

count += 5
data.extend([1, 2, 3, 4, 5])

print(f"✓ Variables modified")
print(f"  count = {count}")
print(f"  data = {data}")
```

## Understanding Execution Order

**Important:** Cells can be run in any order, not just top-to-bottom. But the kernel remembers all variables from cells you have run, regardless of order.

This can cause hidden bugs:
- Cell A works fine when run after Cell B, but fails when Cell B isn't run first
- Your notebook works on your machine but fails for others (they didn't run Cell B)
- Solution: Always **restart** and **run all** before final submission

```python
# Code Cell 3: Use the Modified Variables
# This cell depends on cells 1 and 2 being run first

print(f"Sum of data: {sum(data)}")
print(f"Count value: {count}")
print(f"Total: {sum(data) + count}")
```

## Demonstrating: Long-Running Code (For Interruption)

The cell below runs a loop that takes time. You can:
- **Run it normally** to see it complete
- **Interrupt it** (press the stop button) if it's taking too long

This demonstrates your ability to control execution.

```python
# Code Cell 4: Long-Running Code (Can be Interrupted)
# Run this, then interrupt it to practice kernel control
# Or let it run to completion

import time

print("Starting a 10-second loop (can be interrupted)...")
for i in range(10):
    time.sleep(1)
    print(f"  {i+1} seconds elapsed", end="\r")

print("\n✓ Loop completed successfully")
```

## Restarting the Kernel

When you **restart the kernel**:
- ✅ All variables are cleared from memory
- ✅ Execution counter resets
- ✅ Notebook state is reset to blank
- ❌ All previous outputs disappear

**When to restart:**
- Testing notebook reproducibility (restart + run all)
- Clearing stale variables causing issues
- Starting fresh after debugging
- Before final submission (to ensure notebook works from top to bottom)

**How to restart:** Kernel menu → Restart Kernel → Restart

```python
# Code Cell 5: Verify Kernel State
# After restart: count and data will NOT be defined
# Before restart: they have values from earlier cells

try:
    print(f"count = {count}")
    print(f"data = {data}")
    print("✓ Variables exist (kernel was not restarted)")
except NameError as e:
    print(f"✗ Error: {e}")
    print("Variables are not defined (kernel was just restarted)")
```

## Interrupting vs Restarting

| Action | When to Use | Effect |
|--------|-------------|--------|
| **Interrupt** | Cell is taking too long or infinite loop | Stops execution, kernel stays alive, variables preserved |
| **Restart** | Need clean state, testing reproducibility | Clears all variables, resets kernel |

**Example Scenarios:**
1. Cell is stuck in a 1-hour loop → **Interrupt** to save time
2. Code fails but you're not sure why → **Restart** and run from top
3. Testing if notebook works for someone else → **Restart** + Run All
4. Variables are weird/stale → **Restart** to clear them

```python
# Code Cell 6: Final Kernel Control Check
# After running all cells in order, this confirms kernel management

print("Kernel Management Demonstration Complete")
print("="*50)
print("✓ Cells executed in order")
print("✓ Kernel state preserved variables")
print("✓ Long-running code demonstrated (interruptible)")
print("✓ Kernel restart/interrupt actions understood")
print("="*50)
print("\nBest Practice: Always restart and run-all before final submission!")
```

## Summary

**Key Takeaways:**

1. **The kernel is stateful** - it remembers variables from all cells you've run
2. **Execution order matters** - cells can fail if dependencies aren't run first
3. **Interrupt** stops a running cell without clearing variables
4. **Restart** clears all variables and resets the kernel
5. **Always test with restart + run-all** to ensure reproducibility

**Professional Practice:**
- Run cells intentionally, not randomly
- Restart before final submission
- Document cell dependencies in Markdown
- Use interrupt wisely for long-running code

This ensures your Data Science notebooks are reproducible, debuggable, and ready for review.
