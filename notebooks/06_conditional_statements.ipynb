## 1️⃣ Writing Basic if Statements

A basic **if statement** checks a condition and executes code only when the condition is True.

**Syntax:**
```python
if condition:
    # Code here runs if condition is True
    # Code here is indented (usually 4 spaces)
```

**Important:** Python uses indentation to define code blocks. Incorrect indentation causes errors or unexpected behavior.

**Comparison Operators:**
- `==` Equal to
- `!=` Not equal to
- `<` Less than
- `>` Greater than
- `<=` Less than or equal to
- `>=` Greater than or equal to

````python
# Example 1: Basic if Statement with Numbers
print("=" * 60)
print("BASIC IF STATEMENTS")
print("=" * 60)

age = 18

# Simple if condition
if age >= 18:
    print(f"Age {age}: You are an adult")

print()

# The code below runs regardless (not indented under if)
print("This line always prints")
print()

# Example with False condition
score = 45

if score >= 50:
    print("You passed the test!")

print("This prints even though the if block didn't execute")
print()
````

````python
# Example 2: Different Comparison Operators
print("=" * 60)
print("COMPARISON OPERATORS")
print("=" * 60)

x = 10
y = 20

# Equality
if x == 10:
    print(f"✓ x equals 10: {x == 10}")

# Inequality
if x != y:
    print(f"✓ x is not equal to y: {x != y}")

# Less than
if x < y:
    print(f"✓ x ({x}) is less than y ({y}): {x < y}")

# Greater than
if y > x:
    print(f"✓ y ({y}) is greater than x ({x}): {y > x}")

# Less than or equal to
if x <= 10:
    print(f"✓ x is less than or equal to 10: {x <= 10}")

# Greater than or equal to
if y >= 20:
    print(f"✓ y is greater than or equal to 20: {y >= 20}")

print()

# Example with strings
name = "Alice"

if name == "Alice":
    print(f"Welcome, {name}!")

print()
````

## 2️⃣ Using if-else for Decision Branching

An **if-else statement** provides two paths: one when the condition is True, another when it's False.

**Syntax:**
```python
if condition:
    # Code here runs if condition is True
else:
    # Code here runs if condition is False
```

**Key Points:**
- Only one block executes (either if or else, never both)
- Both paths should handle the logic completely
- Use else to ensure all cases are covered

````python
# Example 3: if-else Decision Branching
print("=" * 60)
print("IF-ELSE STATEMENTS")
print("=" * 60)

# Example 1: Pass/Fail
score = 75

if score >= 60:
    print(f"Score: {score} - PASS ✓")
else:
    print(f"Score: {score} - FAIL ✗")

print()

# Example 2: Login Validation
password = "secret123"
entered_password = "secret123"

if entered_password == password:
    print("Password correct - Access granted! ✓")
else:
    print("Password incorrect - Access denied ✗")

print()

# Example 3: Age verification
age = 16

if age >= 18:
    print(f"Age {age}: You are an adult")
else:
    print(f"Age {age}: You are a minor")

print()

# Example 4: Account balance
balance = 500

if balance > 0:
    print(f"Account balance: ${balance} - ACTIVE")
else:
    print(f"Account balance: ${balance} - SUSPENDED")

print()
````

## 3️⃣ Handling Multiple Conditions with elif

An **if-elif-else statement** allows you to check multiple conditions in sequence.

**Syntax:**
```python
if condition1:
    # Runs if condition1 is True
elif condition2:
    # Runs if condition1 is False and condition2 is True
elif condition3:
    # Runs if condition1 and condition2 are False, and condition3 is True
else:
    # Runs if all conditions are False
```

**Key Points:**
- Python checks conditions from top to bottom
- Only ONE block executes
- Order matters—put most specific conditions first
- Use else as a catch-all for remaining cases

````python
# Example 4: if-elif-else with Multiple Conditions
print("=" * 60)
print("IF-ELIF-ELSE STATEMENTS")
print("=" * 60)

# Example 1: Letter Grade
score = 85

if score >= 90:
    grade = "A"
    print(f"Score {score} → Grade {grade}")
elif score >= 80:
    grade = "B"
    print(f"Score {score} → Grade {grade}")
elif score >= 70:
    grade = "C"
    print(f"Score {score} → Grade {grade}")
elif score >= 60:
    grade = "D"
    print(f"Score {score} → Grade {grade}")
else:
    grade = "F"
    print(f"Score {score} → Grade {grade}")

print()

# Example 2: Transaction Amount Category
amount = 5000

if amount < 100:
    category = "Small"
    print(f"Amount ${amount} → Category: {category} (Instant)")
elif amount < 1000:
    category = "Medium"
    print(f"Amount ${amount} → Category: {category} (Manual Review)")
elif amount < 10000:
    category = "Large"
    print(f"Amount ${amount} → Category: {category} (24-hour Hold)")
else:
    category = "Very Large"
    print(f"Amount ${amount} → Category: {category} (Requires Approval)")

print()

# Example 3: User Status
age = 25

if age < 13:
    status = "Child"
    print(f"Age {age} → Status: {status}")
elif age < 18:
    status = "Teen"
    print(f"Age {age} → Status: {status}")
elif age < 65:
    status = "Adult"
    print(f"Age {age} → Status: {status}")
else:
    status = "Senior"
    print(f"Age {age} → Status: {status}")

print()
````

## 4️⃣ Using Logical Operators

Logical operators combine multiple conditions into one decision.

**The Three Logical Operators:**

1. **and**: Both conditions must be True
   ```python
   if age >= 18 and has_license:
       print("Can drive")
   ```

2. **or**: At least one condition must be True
   ```python
   if is_weekend or is_holiday:
       print("Day off")
   ```

3. **not**: Reverses the condition (True becomes False, False becomes True)
   ```python
   if not is_expired:
       print("Valid")
   ```

**Truth Tables:**

- `True and True` → `True`
- `True and False` → `False`
- `True or False` → `True`
- `False or False` → `False`
- `not True` → `False`
- `not False` → `True`

````python
# Example 5: Using the 'and' Operator
print("=" * 60)
print("LOGICAL OPERATORS: and, or, not")
print("=" * 60)

print("The 'and' operator (both must be True):")
print("-" * 60)

age = 25
has_license = True

# Both conditions must be True
if age >= 18 and has_license:
    print(f"✓ Can drive: age={age}, license={has_license}")
else:
    print(f"✗ Cannot drive: age={age}, license={has_license}")

print()

# Example with False condition
age = 16
has_license = True

if age >= 18 and has_license:
    print(f"✓ Can drive: age={age}, license={has_license}")
else:
    print(f"✗ Cannot drive: age={age}, license={has_license}")

print()

# Example 6: Using the 'or' Operator
print("The 'or' operator (at least one must be True):")
print("-" * 60)

is_weekend = False
is_holiday = True

# At least one condition must be True
if is_weekend or is_holiday:
    print(f"✓ Day off: weekend={is_weekend}, holiday={is_holiday}")
else:
    print(f"✗ Work day: weekend={is_weekend}, holiday={is_holiday}")

print()

# Example with both False
is_weekend = False
is_holiday = False

if is_weekend or is_holiday:
    print(f"✓ Day off: weekend={is_weekend}, holiday={is_holiday}")
else:
    print(f"✗ Work day: weekend={is_weekend}, holiday={is_holiday}")

print()

# Example 7: Using the 'not' Operator
print("The 'not' operator (reverses True/False):")
print("-" * 60)

is_expired = False

if not is_expired:
    print(f"✓ Passport is valid (expired={is_expired})")
else:
    print(f"✗ Passport is invalid (expired={is_expired})")

print()

is_expired = True

if not is_expired:
    print(f"✓ Passport is valid (expired={is_expired})")
else:
    print(f"✗ Passport is invalid (expired={is_expired})")

print()
````

````python
# Example 8: Combining Multiple Logical Operators
print("=" * 60)
print("COMBINING LOGICAL OPERATORS")
print("=" * 60)

# Real-world scenario: Movie eligibility
age = 15
parent_permission = True
completed_form = True

print("Scenario: Movie Watch Eligibility (15-rated film)")
print(f"  Age: {age}, Parent Permission: {parent_permission}, Form: {completed_form}")
print()

if (age >= 15) or (age < 15 and parent_permission and completed_form):
    print("✓ Can watch the movie")
else:
    print("✗ Cannot watch the movie")

print()

# Real-world scenario: Credit approval
credit_score = 680
annual_income = 45000
debt_to_income = 0.35

print("Scenario: Credit Card Approval")
print(f"  Credit Score: {credit_score}")
print(f"  Annual Income: ${annual_income:,}")
print(f"  Debt-to-Income Ratio: {debt_to_income:.1%}")
print()

if (credit_score >= 650) and (annual_income >= 30000) and (debt_to_income < 0.5):
    print("✓ APPROVED")
else:
    print("✗ DENIED")

print()

# Real-world scenario: Fraud detection
amount = 2500
is_unusual_location = True
is_late_night = True

print("Scenario: Transaction Fraud Detection")
print(f"  Amount: ${amount}")
print(f"  Unusual Location: {is_unusual_location}")
print(f"  Late Night: {is_late_night}")
print()

if (amount > 2000) and (is_unusual_location or is_late_night):
    print("⚠️  FLAGGED FOR REVIEW (Potential fraud)")
elif amount > 5000:
    print("⚠️  FLAGGED FOR REVIEW (Large amount)")
else:
    print("✓ APPROVED")

print()
````

## 5️⃣ Common Mistakes and Debugging

### Common Mistake #1: Using = instead of ==

```python
# ❌ WRONG: Assignment instead of comparison
if x = 5:
    print("x is 5")

# ✅ CORRECT: Use == for comparison
if x == 5:
    print("x is 5")
```

### Common Mistake #2: Type Mismatch

```python
# ❌ WRONG: Comparing string to number
user_input = "25"  # String from input
if user_input > 18:  # This causes unexpected behavior
    print("Adult")

# ✅ CORRECT: Convert to proper type first
user_input = int(user_input)  # Convert to int
if user_input > 18:
    print("Adult")
```

### Common Mistake #3: Indentation Errors

```python
# ❌ WRONG: Incorrect indentation
if age >= 18:
print("Adult")  # Not indented - causes error

# ✅ CORRECT: Proper indentation
if age >= 18:
    print("Adult")  # Indented correctly
```

### Common Mistake #4: Condition Always False

```python
# ❌ WRONG: Impossible condition
if x > 10 and x < 5:
    print("This never runs")  # x cannot be > 10 AND < 5

# ✅ CORRECT: Use or for multiple ranges
if x < 5 or x > 10:
    print("x is outside 5-10 range")
```

### Debugging Tip: Add Print Statements

```python
x = 8
print(f"x = {x}")
print(f"x > 5: {x > 5}")
print(f"x < 10: {x < 10}")
print(f"x > 5 and x < 10: {x > 5 and x < 10}")

if x > 5 and x < 10:
    print("Between 5 and 10")
```

````python
# Example 9: Debugging Conditions with Print Statements
print("=" * 60)
print("DEBUGGING CONDITIONAL LOGIC")
print("=" * 60)

# Scenario: Check if value is in valid range
value = 7

print(f"Value: {value}")
print(f"  value > 5: {value > 5}")
print(f"  value < 10: {value < 10}")
print(f"  value > 5 and value < 10: {value > 5 and value < 10}")
print()

if value > 5 and value < 10:
    print(f"✓ {value} is between 5 and 10")
else:
    print(f"✗ {value} is NOT between 5 and 10")

print()

# Example: Type mismatch debugging
print("Type Mismatch Example:")
print("-" * 60)

age_input = "25"  # From user input (string)
print(f"age_input = '{age_input}' (type: {type(age_input).__name__})")

# ❌ Without conversion
try:
    if age_input > 18:
        print("Adult")
except TypeError as e:
    print(f"❌ Error: Cannot compare string with int directly")

print()

# ✅ With conversion
age = int(age_input)
print(f"age = {age} (type: {type(age).__name__})")

if age > 18:
    print(f"✓ Age {age}: Adult")
else:
    print(f"✓ Age {age}: Minor")

print()
````

## 📊 Summary & Quick Reference

### Conditional Statement Structure

```python
if condition:
    # Only runs if condition is True
elif condition2:
    # Only runs if condition is False and condition2 is True
elif condition3:
    # Only runs if both previous conditions are False and this is True
else:
    # Runs if all conditions are False
```

### Comparison Operators

| Operator | Meaning | Example |
|----------|---------|---------|
| `==` | Equal to | `x == 5` |
| `!=` | Not equal to | `x != 5` |
| `<` | Less than | `x < 5` |
| `>` | Greater than | `x > 5` |
| `<=` | Less than or equal | `x <= 5` |
| `>=` | Greater than or equal | `x >= 5` |

### Logical Operators

| Operator | Meaning | Example |
|----------|---------|---------|
| `and` | Both must be True | `age >= 18 and has_license` |
| `or` | At least one must be True | `is_weekend or is_holiday` |
| `not` | Reverses True/False | `not is_expired` |

### Decision Tree Logic

```
if condition1 is True:
    → Execute block 1
else if condition2 is True:
    → Execute block 2
else if condition3 is True:
    → Execute block 3
else:
    → Execute block 4
```

Only ONE block executes. Python evaluates top to bottom and stops at the first True condition.

### Key Rules

1. **Indentation matters** - Code inside if/elif/else must be indented
2. **Use == for comparison** - Not = (which is assignment)
3. **Order conditions carefully** - Check specific conditions before general ones
4. **Ensure both paths are valid** - If using if-else, both should handle the logic
5. **Use and/or to combine conditions** - Keep combined conditions readable
6. **Add print statements for debugging** - Verify what values are being evaluated

### Common Use Cases

**Validation:**
```python
if age >= 18:
    print("Adult")
```

**Authentication:**
```python
if password == correct_password:
    print("Access granted")
```

**Range checking:**
```python
if 0 <= score <= 100:
    print("Valid score")
```

**Multiple criteria:**
```python
if amount > 1000 and not is_verified:
    print("Review required")
```

Conditional logic is the backbone of intelligent programs. Mastering it enables you to write code that makes decisions based on data!
