## 1️⃣ Working with Python Lists

**Lists** are ordered, mutable collections enclosed in square brackets `[]`. You can:
- Store multiple values in a single variable
- Access elements using their index (starting from 0)
- Modify, add, and remove elements
- Lists are perfect for dynamic data that changes

**Key Characteristics:**
- **Ordered**: Elements have a fixed position
- **Mutable**: You can change, add, or remove elements
- **Indexed**: Access elements by position [0], [1], etc.
- **Flexible**: Can store different types of data

````python
# Example 1: Creating and Accessing Lists
print("=" * 60)
print("LISTS: Creating and Accessing Elements")
print("=" * 60)

# Create lists
fruits = ["apple", "banana", "cherry", "date"]
numbers = [10, 20, 30, 40, 50]
mixed = [42, "hello", 3.14, True]

print("List 1 - Fruits:")
print(f"  {fruits}")
print()

print("List 2 - Numbers:")
print(f"  {numbers}")
print()

print("List 3 - Mixed types:")
print(f"  {mixed}")
print()

# Accessing elements using index (0-based)
print("Accessing elements by index:")
print("-" * 60)
print(f"First fruit (index 0): {fruits[0]}")
print(f"Second fruit (index 1): {fruits[1]}")
print(f"Last fruit (index -1): {fruits[-1]}")
print(f"Second-to-last fruit (index -2): {fruits[-2]}")
print()

# Length of list
print(f"Number of fruits: {len(fruits)}")
print()
````

````python
# Example 2: Modifying Lists
print("=" * 60)
print("LISTS: Modifying Elements")
print("=" * 60)

# Create a list
colors = ["red", "green", "blue"]
print(f"Original list: {colors}")

# Modify an element (replace)
colors[1] = "yellow"
print(f"After changing index 1: {colors}")

# Add an element to the end
colors.append("purple")
print(f"After appending 'purple': {colors}")

# Insert an element at a specific position
colors.insert(0, "orange")
print(f"After inserting 'orange' at index 0: {colors}")

# Remove an element by value
colors.remove("yellow")
print(f"After removing 'yellow': {colors}")

# Remove an element by index
removed = colors.pop(2)
print(f"After popping index 2 (removed '{removed}'): {colors}")

# Remove all elements
copy_list = colors.copy()
copy_list.clear()
print(f"After clearing: {copy_list}")
print()
````

## 2️⃣ Working with Python Tuples

**Tuples** are ordered, immutable collections enclosed in parentheses `()`. You can:
- Store multiple fixed values in a single variable
- Access elements using their index
- **Cannot modify, add, or remove elements** after creation
- Perfect for data that should never change

**Key Characteristics:**
- **Ordered**: Elements have a fixed position
- **Immutable**: You cannot change elements (prevents accidents)
- **Indexed**: Access elements by position like lists
- **Protective**: Great for configuration values and constants
- **Hashable**: Can be used as dictionary keys (lists cannot)

````python
# Example 3: Creating and Accessing Tuples
print("=" * 60)
print("TUPLES: Creating and Accessing Elements")
print("=" * 60)

# Create tuples
coordinates = (40.7128, -74.0060)  # Latitude, Longitude of NYC
rgb_color = (255, 0, 128)          # RGB color values
mixed_tuple = (42, "hello", 3.14)  # Mixed types

print("Tuple 1 - Coordinates (lat, lon):")
print(f"  {coordinates}")
print()

print("Tuple 2 - RGB Color values:")
print(f"  {rgb_color}")
print()

print("Tuple 3 - Mixed types:")
print(f"  {mixed_tuple}")
print()

# Accessing elements using index (same as lists)
print("Accessing elements by index:")
print("-" * 60)
print(f"First coordinate (latitude): {coordinates[0]}")
print(f"Second coordinate (longitude): {coordinates[1]}")
print(f"Red channel: {rgb_color[0]}")
print(f"Green channel: {rgb_color[1]}")
print(f"Blue channel: {rgb_color[2]}")
print()

# Length of tuple
print(f"Number of elements in coordinates: {len(coordinates)}")
print()
````

````python
# Example 4: Immutability - The Strength of Tuples
print("=" * 60)
print("TUPLES: Immutability (Cannot Modify)")
print("=" * 60)

# Create a tuple
settings = ("localhost", 8080, "development")
print(f"Original settings: {settings}")
print(f"  Host: {settings[0]}")
print(f"  Port: {settings[1]}")
print(f"  Environment: {settings[2]}")
print()

# Try to modify a tuple element
print("Attempting to modify tuple (will cause error):")
print("-" * 60)
try:
    settings[1] = 9000  # Try to change port
except TypeError as error:
    print(f"❌ ERROR: {error}")
print()

# Try to add to a tuple
print("Attempting to add to tuple (will cause error):")
print("-" * 60)
try:
    settings.append("debug")  # Tuples don't have append()
except AttributeError as error:
    print(f"❌ ERROR: {error}")
print()

# The proper way: create a new tuple
print("✅ PROPER WAY: Create a new tuple if you need different values")
print("-" * 60)
new_settings = ("localhost", 9000, "production")
print(f"New settings: {new_settings}")
print()

print("KEY INSIGHT:")
print("Tuples protect your data from accidental modification.")
print("If you need to change values, you create a new tuple.")
print()
````

## 3️⃣ Working with Python Dictionaries

**Dictionaries** are unordered, mutable collections with key-value pairs enclosed in curly braces `{}`. You can:
- Store data with meaningful keys (not just numeric indexes)
- Access values using their key
- Modify, add, and remove key-value pairs
- Perfect for modeling real-world entities

**Key Characteristics:**
- **Key-Value Pairs**: Each value has a meaningful key
- **Mutable**: You can change, add, or remove pairs
- **Keyed Access**: Use keys instead of indexes
- **Unordered**: Order doesn't matter (Python 3.7+ maintains insertion order)
- **Structured**: Great for representing objects like user profiles, configuration

````python
# Example 5: Creating and Accessing Dictionaries
print("=" * 60)
print("DICTIONARIES: Creating and Accessing Values")
print("=" * 60)

# Create dictionaries
person = {
    "name": "Alice Johnson",
    "age": 28,
    "email": "alice@example.com",
    "city": "San Francisco"
}

book = {
    "title": "Python Basics",
    "author": "John Smith",
    "year": 2020,
    "pages": 350
}

print("Dictionary 1 - Person:")
print(f"  {person}")
print()

print("Dictionary 2 - Book:")
print(f"  {book}")
print()

# Accessing values using keys (NOT indexes!)
print("Accessing values using keys:")
print("-" * 60)
print(f"Name: {person['name']}")
print(f"Email: {person['email']}")
print(f"City: {person['city']}")
print()

print(f"Book title: {book['title']}")
print(f"Author: {book['author']}")
print(f"Published: {book['year']}")
print()

# Using .get() for safer access
print("Safer access with .get() (returns None if key doesn't exist):")
print(f"Phone: {person.get('phone')}")
print(f"Phone with default: {person.get('phone', 'N/A')}")
print()

# Number of key-value pairs
print(f"Number of fields in person: {len(person)}")
print()
````

````python
# Example 6: Modifying Dictionaries
print("=" * 60)
print("DICTIONARIES: Modifying Key-Value Pairs")
print("=" * 60)

# Create a user profile
user = {
    "username": "alice_wonder",
    "email": "alice@example.com",
    "age": 28
}
print(f"Original user: {user}")
print()

# Modify an existing key
user["age"] = 29
print(f"After birthday update: {user}")

# Add a new key-value pair
user["phone"] = "555-1234"
print(f"After adding phone: {user}")

# Add another new pair
user["verified"] = True
print(f"After adding verified status: {user}")

# Remove a key-value pair
removed = user.pop("phone")
print(f"After removing phone (removed '{removed}'): {user}")

# Update multiple values at once
user.update({"age": 30, "city": "Los Angeles"})
print(f"After update: {user}")

# Remove all pairs
test_dict = {"a": 1, "b": 2}
test_dict.clear()
print(f"After clearing: {test_dict}")

print()
````

## 4️⃣ Choosing the Right Data Structure

Different situations call for different structures. Here's how to choose:

### When to Use LISTS
- ✅ You have a sequence of items (items, scores, names)
- ✅ Order matters
- ✅ You need to frequently add, remove, or modify elements
- ✅ You access items by position

**Example**: `students = ["Alice", "Bob", "Charlie"]`

### When to Use TUPLES
- ✅ You have fixed values that should never change
- ✅ You need to protect data from accidental modification
- ✅ You're using as dictionary keys or function arguments
- ✅ Data represents coordinates, RGB values, or fixed configurations

**Example**: `color = (255, 128, 0)` # RGB doesn't change

### When to Use DICTIONARIES
- ✅ You have labeled data with meaningful keys
- ✅ You need to look up values by name (not position)
- ✅ You're modeling a real-world entity (user, product, settings)
- ✅ You frequently add or remove attributes

**Example**: `user = {"name": "Alice", "age": 28, "email": "alice@example.com"}`

````python
# Example 7: Real-World Scenario - Student Information
print("=" * 60)
print("SCENARIO: Storing Student Information")
print("=" * 60)

# ❌ BAD: Using list (hard to understand what each number means)
student_list = ["Alice", 28, "alice@example.com", "San Francisco"]
print("❌ Using LIST (unclear what each position means):")
print(f"  {student_list}")
print(f"  student_list[2] = {student_list[2]}  # What does this represent?")
print()

# ✅ GOOD: Using dictionary (clear, self-documenting)
student_dict = {
    "name": "Alice",
    "age": 28,
    "email": "alice@example.com",
    "city": "San Francisco"
}
print("✅ Using DICTIONARY (clear and self-documenting):")
print(f"  {student_dict}")
print(f"  student_dict['email'] = {student_dict['email']}  # Clear meaning!")
print()
````

````python
# Example 8: Lists vs Tuples - Data Safety
print("=" * 60)
print("SCENARIO: Configuration Values (Should Not Change)")
print("=" * 60)

# ❌ RISKY: Using list for configuration (can be accidentally modified)
config_list = ["localhost", 8080, "development"]
print("❌ Using LIST for configuration (risky - can be modified):")
print(f"  Original: {config_list}")

# Someone accidentally modifies it
config_list[1] = 9000  # BUG: Port changed!
config_list[2] = "production"  # BUG: Environment changed!
print(f"  After accidental changes: {config_list}")
print(f"  😞 Configuration corrupted!")
print()

# ✅ SAFE: Using tuple for configuration (cannot be modified)
config_tuple = ("localhost", 8080, "development")
print("✅ Using TUPLE for configuration (safe - cannot be modified):")
print(f"  Original: {config_tuple}")

# If we try to modify it:
try:
    config_tuple[1] = 9000
except TypeError:
    print(f"  ❌ Cannot modify - TypeError raised!")
    print(f"  ✅ Configuration protected!")
    print(f"  Current: {config_tuple}")

print()

print("KEY LESSON:")
print("Use TUPLE for values that should NEVER change (configuration, constants)")
print("Use LIST for values that change over time (student list, transactions)")
print()
````

````python
# Example 9: Combining All Three Structures
print("=" * 60)
print("COMBINED EXAMPLE: Real Database Record")
print("=" * 60)

# A complete record combining lists, tuples, and dictionaries
users_database = [
    {
        "id": 1,
        "name": "Alice Johnson",
        "email": "alice@example.com",
        "coordinates": (40.7128, -74.0060),  # (latitude, longitude)
        "tags": ["student", "verified"],
        "settings": {"notifications": True, "theme": "dark"}
    },
    {
        "id": 2,
        "name": "Bob Smith",
        "email": "bob@example.com",
        "coordinates": (34.0522, -118.2437),
        "tags": ["student"],
        "settings": {"notifications": False, "theme": "light"}
    }
]

print("Database with 2 users:")
print()

# Access nested data
for i, user in enumerate(users_database, 1):
    print(f"User {i}:")
    print(f"  Name: {user['name']}")
    print(f"  Email: {user['email']}")
    print(f"  Location (lat, lon): {user['coordinates']}")
    print(f"  Tags: {user['tags']}")
    print(f"  Dark theme? {user['settings']['theme'] == 'dark'}")
    print()

print("KEY INSIGHT:")
print("Real data is often complex and nested!")
print("Lists hold multiple records")
print("Dictionaries model individual records")
print("Tuples protect fixed data like coordinates")
print("Lists within dictionaries store variable attributes")
print()
````

## 📊 Summary & Quick Reference

### Comparison Table

| Feature | List | Tuple | Dictionary |
|---------|------|-------|-----------|
| **Syntax** | `[1, 2, 3]` | `(1, 2, 3)` | `{"a": 1, "b": 2}` |
| **Mutable** | ✅ Yes | ❌ No | ✅ Yes |
| **Access** | Index: `[0]` | Index: `[0]` | Key: `["key"]` |
| **Order** | ✅ Ordered | ✅ Ordered | ✅ Ordered (Python 3.7+) |
| **Use Case** | Sequences | Fixed data | Labeled data |
| **Protection** | Low | High | Medium |

### Common Operations

**Lists:**
```python
fruits = ["apple", "banana"]
fruits.append("cherry")        # Add element
fruits.remove("banana")        # Remove by value
item = fruits.pop(0)           # Remove by index
fruits[0] = "orange"           # Modify element
```

**Tuples:**
```python
coordinates = (40.71, -74.00)
lat = coordinates[0]           # Access by index
# Cannot modify - data is protected!
```

**Dictionaries:**
```python
user = {"name": "Alice", "age": 28}
user["email"] = "alice@example.com"  # Add/modify
age = user["age"]              # Access by key
del user["age"]                # Remove key
```

### When to Use What

**LIST**: Dynamic sequences where items change
- Shopping lists
- Student scores
- Transaction history

**TUPLE**: Fixed, protected data
- Configuration values
- Function return values (multiple values)
- Coordinates or RGB colors

**DICTIONARY**: Labeled data representing objects
- User profiles
- Configuration settings
- Database records

### Critical Rules

1. **Lists are flexible** - Use when data changes
2. **Tuples protect** - Use for fixed, important data
3. **Dicts are clear** - Use meaningful keys instead of numeric indexes
4. **Nest carefully** - List of dicts is common and powerful
5. **Choose intentionally** - Structure choice affects code quality

Mastering collections is essential for all Python and Data Science work!
