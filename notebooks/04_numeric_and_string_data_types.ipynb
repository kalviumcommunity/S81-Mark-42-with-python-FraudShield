## 1️⃣ Working with Numeric Data Types

Python represents numbers in two main ways:
- **Integers (int)**: Whole numbers without decimal points (e.g., 42, -10, 0)
- **Floats (float)**: Numbers with decimal points (e.g., 3.14, -2.5, 0.0)

Let's explore numeric data types and basic arithmetic operations.

```python
# Example 1: Creating Integer Variables
print("=" * 50)
print("INTEGERS (Whole Numbers)")
print("=" * 50)

# Create integer variables
age = 25
score = 100
temperature = -5
count = 0

# Print the variables
print(f"Age: {age}")
print(f"Score: {score}")
print(f"Temperature: {temperature}")
print(f"Count: {count}")
print()
```

```python
# Example 2: Creating Floating-Point Variables
print("=" * 50)
print("FLOATS (Decimal Numbers)")
print("=" * 50)

# Create float variables
price = 19.99
pi = 3.14159
percentage = 85.5
balance = -10.50

# Print the variables
print(f"Price: {price}")
print(f"Pi: {pi}")
print(f"Percentage: {percentage}")
print(f"Balance: {balance}")
print()
```

```python
# Example 3: Basic Arithmetic Operations
print("=" * 50)
print("ARITHMETIC OPERATIONS")
print("=" * 50)

# Define numbers for calculations
a = 10
b = 3
c = 2.5

# Addition
result_add = a + b
print(f"Addition: {a} + {b} = {result_add}")

# Subtraction
result_sub = a - b
print(f"Subtraction: {a} - {b} = {result_sub}")

# Multiplication
result_mul = a * b
print(f"Multiplication: {a} * {b} = {result_mul}")

# Division (important: always returns float)
result_div = a / b
print(f"Division: {a} / {b} = {result_div}")
print(f"  Type of result: {type(result_div)}")

# Floor Division (returns integer result)
result_floor = a // b
print(f"Floor Division: {a} // {b} = {result_floor}")
print(f"  Type of result: {type(result_floor)}")

# Modulo (remainder)
result_mod = a % b
print(f"Modulo (remainder): {a} % {b} = {result_mod}")

# Power/Exponentiation
result_pow = a ** 2
print(f"Power: {a} ** 2 = {result_pow}")

print()
```

## 2️⃣ Understanding String Data Types

Strings represent text and character data in Python. Strings are enclosed in quotes (single `'`, double `"`, or triple `"""`)

Key string operations:
- **Concatenation**: Joining strings together with `+`
- **Access**: Getting individual characters with indexing `[index]`
- **Formatting**: Inserting values into strings with f-strings or `.format()`
- **Length**: Finding string length with `len()`

```python
# Example 4: Creating String Variables
print("=" * 50)
print("STRINGS (Text Data)")
print("=" * 50)

# Create string variables
first_name = "John"
last_name = "Doe"
city = "New York"
message = "Hello, World!"

print(f"First Name: {first_name}")
print(f"Last Name: {last_name}")
print(f"City: {city}")
print(f"Message: {message}")
print()
```

```python
# Example 5: String Concatenation
print("=" * 50)
print("STRING CONCATENATION (Joining Strings)")
print("=" * 50)

# Using + operator
full_name = first_name + " " + last_name
print(f"Using +: {full_name}")

# Using f-strings (recommended)
greeting = f"Welcome, {first_name} {last_name}!"
print(f"Using f-string: {greeting}")

# Using .format()
location = "Hello {}, you are from {}".format(first_name, city)
print(f"Using .format(): {location}")

print()
```

```python
# Example 6: String Indexing and Length
print("=" * 50)
print("ACCESSING STRINGS & LENGTH")
print("=" * 50)

text = "Python"

# Accessing individual characters (0-indexed)
print(f"Text: {text}")
print(f"First character: {text[0]}")  # P
print(f"Second character: {text[1]}")  # y
print(f"Third character: {text[2]}")   # t
print(f"Last character: {text[-1]}")   # n

# String length
length = len(text)
print(f"Length of '{text}': {length}")

# Slicing (getting substrings)
print(f"First 2 characters: {text[0:2]}")   # Py
print(f"Last 3 characters: {text[-3:]}")    # hon

print()
```

## 3️⃣ Mixing Numbers and Strings Safely

A common problem occurs when you accidentally treat numbers as strings or vice versa. Python will raise an error if you try to mix incompatible types without explicit conversion.

```python
# Example 7: Type Mismatch - What NOT to Do
print("=" * 50)
print("TYPE MISMATCH ERRORS")
print("=" * 50)

# Problem 1: Cannot add number and string
price = 19.99
quantity_str = "5"  # This is a STRING, not a number

print(f"Price: {price} (type: {type(price)})")
print(f"Quantity: {quantity_str} (type: {type(quantity_str)})")

# This will cause an error:
try:
    total = price * quantity_str  # ❌ ERROR: Can't multiply float and str
except TypeError as error:
    print(f"❌ ERROR: {error}")

print()
```

```python
# Example 8: Type Conversion - The Solution
print("=" * 50)
print("TYPE CONVERSION (Converting Between Types)")
print("=" * 50)

# Solution: Convert string to number using int() or float()
price = 19.99
quantity_str = "5"

# Convert string to integer
quantity = int(quantity_str)  # ✅ Convert string to int

# Now we can calculate
total = price * quantity
print(f"Price: ${price}")
print(f"Quantity: {quantity}")
print(f"Total: ${total:.2f}")

print()

# Example: Converting user input
print("Converting different types:")
print("-" * 50)

# String to integer
number_str = "42"
number_int = int(number_str)
print(f"String '{number_str}' → Integer: {number_int}")

# String to float
decimal_str = "3.14"
decimal_float = float(decimal_str)
print(f"String '{decimal_str}' → Float: {decimal_float}")

# Integer to string
age = 25
age_str = str(age)
print(f"Integer {age} → String: '{age_str}'")

# Float to integer (truncates decimal)
price = 19.99
price_int = int(price)
print(f"Float {price} → Integer: {price_int}")

print()
```

## 4️⃣ Inspecting Data Types

Use the `type()` function to check the data type of any variable. This is crucial for debugging and understanding your data.

```python
# Example 9: Using type() to Inspect Variables
print("=" * 50)
print("INSPECTING DATA TYPES WITH type()")
print("=" * 50)

# Different variables
count = 42
rating = 4.5
name = "Alice"
is_active = True

# Check their types
print(f"Variable: count = {count}")
print(f"  Type: {type(count)}")
print()

print(f"Variable: rating = {rating}")
print(f"  Type: {type(rating)}")
print()

print(f"Variable: name = '{name}'")
print(f"  Type: {type(name)}")
print()

print(f"Variable: is_active = {is_active}")
print(f"  Type: {type(is_active)}")
print()
```
